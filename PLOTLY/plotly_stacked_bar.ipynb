# Stacked Bar Chart

```python
import plotly.express as px
import pandas as pd
import os

# HTML Full Output Path
output_filename = "plotly_stacked_bar_chart.html"
full_path = os.path.join(os.getcwd(), output_filename)

# 1. Create a simple dataset
data = {
    "City": ["New York", "New York", "New York", "London", "London", "London"],
    "Product": ["Apples", "Oranges", "Bananas", "Apples", "Oranges", "Bananas"],
    "Sales": [400, 300, 500, 250, 600, 300]
}

color_map = {
    "Apples": "red",            # CSS color name
    "Oranges": "#FFA500",       # Hex code
    "Bananas": "rgb(255, 225, 53)" # RGB string
}

df = pd.DataFrame(data)

# 2. Create the stacked bar chart
# 'x' is the category on the axis
# 'y' is the numerical value
# 'color' tells Plotly how to stack the bars
fig = px.bar(df, 
             x="City", 
             y="Sales", 
             color="Product", 
             color_discrete_map=color_map, # Apply the custom colors here
             title="Fruit Sales by City (Stacked)",
             labels={"Sales": "Total Revenue ($)"},
             template="plotly_white")

# 3. Show the chart
fig.show()
fig.write_html(full_path)
```
