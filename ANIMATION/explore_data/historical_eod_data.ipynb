# Historical End of Day Stock Data
Data has an entry for every stock close from Jan 2006 through Dec 1017

## Read raw data describing stock ticker, sector and company name

```python
import pandas as pd
import io # need this to get info
import plotly.express as px

df_historical_stock:pd.DataFrame = None

df_historical_stock = pd.read_csv('../../data/all_stocks_2006-01-01_to_2018-01-01.csv',encoding='utf8', comment='#')

# Test if empty
print('empty: \n', df_historical_stock.empty)
print('*' *50)

# View range of years - this is data per year
df_historical_stock['Year'] = pd.to_datetime(df_historical_stock['Date']).dt.year
year_range = df_historical_stock.Year.max() - df_historical_stock.Year.min()
print(f"The range of years is: {year_range}, min:{df_historical_stock.Year.min()}, max: {df_historical_stock.Year.max()}")

print('*' *50)

# View info
print('\ninfo: \n')
buffer = io.StringIO()
df_historical_stock.info(buf=buffer)
s = buffer.getvalue()
print(s)
print('*' *50)

# View Line Chart of Stock MMM open value over time
print('\nLine Chart of Stock MMM open value over time:\n')
# Plot a line chart of open values for stock MMM over time
df_mmm = df_historical_stock[df_historical_stock['Name'] == 'MMM']
fig = px.line(
    df_mmm, 
    x='Open', 
    y='Year')
fig.show()
```

## View missing data

```python
# View missing data
# Returns True if any null values exist
print('Is there missing data? \n', df_historical_stock.isnull().any().any())
print('*' *50)

total_missing = df_historical_stock.isnull().sum().sum()
print(f"Total number of missing values: {total_missing}")

missing_per_column = df_historical_stock.isnull().sum()
print("\nMissing values per column:\n", missing_per_column)
```

## Average Close Per Year

```python
import pandas as pd
import plotly.express as px


# Sample data (replace this with your actual data loading)
df = pd.read_csv('../../data/all_stocks_2006-01-01_to_2018-01-01.csv',encoding='utf8',comment='#')

# Ensure 'Date' is in datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Extract the year into a new column
df['Year'] = df['Date'].dt.year

# Group by 'Year' and calculate average Close
df_yearly_avg = df.groupby('Year', as_index=False)['Close'].mean()

# Group by 'Date' and calculate average Close
# df_avg = df.groupby('Date', as_index=False)['Close'].mean()

# Rename 'Close' to 'Avg_Close'
df_yearly_avg.rename(columns={'Close': 'Avg_Close'}, inplace=True)

# The resulting df_avg will have datetime-formatted 'Date'
print(df_yearly_avg)

fig = px.line(df_yearly_avg,x='Year', y='Avg_Close')
fig.show()
```
