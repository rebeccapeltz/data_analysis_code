# Stock Ticker, Sector, Company Name

## Read raw data describing stock ticker, sector and company name

```python
import pandas as pd
import io # need this to get info

df_stock_sector_company = pd.read_csv('../../data/stock_sector_company.csv',comment='#',encoding='utf8')

# Test if empty
print('empty: \n', df_stock_sector_company.empty)
print('*' *50)


# View info
print('info: \n', df_stock_sector_company.info())
print('*' *50)
```

## View Missingn Data

```python
# View missing data
# Returns True if any null values exist
print('Is there missing data? \n', df_stock_sector_company.isnull().any().any())
print('*' *50)

total_missing = df_stock_sector_company.isnull().sum().sum()
print(f"Total number of missing values: {total_missing}")

missing_per_column = df_stock_sector_company.isnull().sum()
print("\nMissing values per column:\n", missing_per_column)
```

## View sample of all rows of date sorted by stock ticker

```python

# View a random sample of all rows of data sorted by stock ticker
number_stocks = len(df_stock_sector_company)
print('number of stocks: ',number_stocks)
print("Sample includes all data")
# sample and sort data
print('sort: \n', df_stock_sector_company.sample(number_stocks).sort_values(by='company'))
print('*' *50)
```
