# C02 Emissions Data Exploration

## Read and Report on C02 Emissions Raw Data

```python
import pandas as pd
import io # need this to get info
import plotly.express as px

# Declare DataFrame
df_emissions: pd.DataFrame = None

df_emissions = pd.read_csv('../../data/c02_emissions.csv',comment='#',header=0, encoding='utf8')

# Test if empty dataframe
print('\n is dataset empty: \n', df_emissions.empty)
print('*' *50)

# View range of years - this is data per year
year_range = df_emissions.Year.max() - df_emissions.Year.min()
print(f"The range of years is: {year_range}, min:{df_emissions.Year.min()}, max: {df_emissions.Year.max()}")

# View info
print('\ninfo: \n')
buffer = io.StringIO()
df_emissions.info(buf=buffer)
s = buffer.getvalue()
print(s)
print('*' *50)

# View Line Chart of Total Emissions Over Time
print('\nLine Chart of Total Emissions Over Time')
# Create a new column for total emissions to create short axis label
df_emissions['Total_Emissions'] = df_emissions['Total carbon emissions from fossil fuel consumption and cement production (million metric tons of C)']  

# Plot a line chart of total carbon emissions over the years
fig = px.line(
    df_emissions, 
    x='Year', 
    y='Total_Emissions')
fig.show()
```

## View missing data

```python
# View missing data
# Returns True if any null values exist
print('Is there missing data? \n', df_emissions.isnull().any().any())
print('*' *50)

# count the numb of missing values in each column
total_missing = df_emissions.isnull().sum()
print(f"Total number of missing values:\n{total_missing}")

missing_per_column = df_emissions.isnull().sum().sum()
print("\nMissing values per column:\n", missing_per_column)
```
