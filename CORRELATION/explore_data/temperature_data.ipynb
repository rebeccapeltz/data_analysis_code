# NOAA Temperature Anomolies

## Read raw data from data downloaded from NOAA

```python
import pandas as pd
import io # need this to get info
import plotly.express as px

# Declare DataFrame
df_temp_anomalies:pd.DataFrame = None

df_temp_anomalies = pd.read_csv('../../data/global_temp_anomaly.csv',comment='#',encoding='utf8')

# Test if empty dataframe
print('\n is dataset empty: \n', df_temp_anomalies.empty)
print('*' *50)

# View range of years - this is data per year
year_range = df_temp_anomalies.Year.max() - df_temp_anomalies.Year.min()
print(f"The range of years is: {year_range}, min:{df_temp_anomalies.Year.min()}, max: {df_temp_anomalies.Year.max()}")

# View info
print('\ninfo: \n')
buffer = io.StringIO()
df_temp_anomalies.info(buf=buffer)
s = buffer.getvalue()
print(s)
print('*' *50)

# View Line Chart of Temperature Anomalies Over Time
print('\nLine Chart of Temperature Anomalies Over Time')
# Plot a line chart of temperature anomalies over the years
fig = px.line(
    df_temp_anomalies, 
    x='Year', 
    y='Anomaly')
fig.show()
```

## View missing data

```python
# View missing data
# Returns True if any null values exist
print('Is there missing data? \n', df_temp_anomalies.isnull().any().any())
print('*' *50)

total_missing = df_temp_anomalies.isnull().sum().sum()
print(f"Total number of missing values: {total_missing}")

missing_per_column = df_temp_anomalies.isnull().sum()
print("\nMissing values per column:\n", missing_per_column)
```
