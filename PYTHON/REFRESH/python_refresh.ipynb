## Variables and Types
### Strings, Integers, Floats, and Boolean

```python
# even though your provide a type, Python allows you to change it without error

# String
year: str = "2026"
print(year, type(year))

# Integer
year: int = 2026
print(year, type(year))

# Float
price: float = 1.25
print(price, type(price))

# Boolean
the_sun_rises_in_the_east: bool = True
the_sun_sets_in_the_east: bool = False
print("the sun rises in the east:", the_sun_rises_in_the_east, type(the_sun_rises_in_the_east))
print("the sun sets in the east:", the_sun_sets_in_the_east, type(the_sun_sets_in_the_east))
```

## Strings and F strings

```python
newline = '\n'
# strings hold text
# using an escape character to allow a single quote within the text of string
famous_quote = 'It always seems impossible until it\'s done. (Nelson Mandela)'
print('\n\n', famous_quote,'\n')

# f strings "interpolate" variables
author = 'Plato'
print(newline,f'The greatest wealth is to live content with little. ({author})')

```

## Special Characters

```python
# newline

print('line 1\nline 2')

# tab

print('section1\tsection 2')
```

## Functions
- use Datetime to get current time
- create function to get current time

```python
from datetime import datetime

# function to get current time
def get_current_time():
    return datetime.now()

# call function    
current_time = get_current_time()
print(f'Current time: {current_time}')

# print text and data using fstring
print(f'Current time: {get_current_time()}')
```
