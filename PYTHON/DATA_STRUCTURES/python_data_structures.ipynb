## Lists  
Data indexed by number starting at 0

````python
# printing a list will include square brackets and comma separated values
print("names list: ", names,"\n")

# lists entries are indexed starting at 0
names = ['bob','sue','tom','mary']
print("first name in list: ", names[0],"\n")
print("last name in list: ", names[3],"\n")

# USING THE COLON to access values in list
# The value before the colon is included in the output, and the value after the colon is excluded from the output.
# If the value before the colon is not specified, the output includes the first item in the list.
# If the value after the colon is not specified, the output includes the last item in the list.

# names[:] -- all values in list 
print("names[:] -- all names starting at position 0:\n", names[:],"\n")

# names[2:] -- all values starting with index #2 to end of string
print("names[2:] -- all names to end of list starting at position 2:\n", names[2:],"\n")

# names[:3] -- all values starting at the beginning of list not including the ending index
print("names[:3] -- all names starting at position 0 and NOT including position 3 or greater:\n", names[:3],"\n")

# names[1:3] -- all values starting at first index and not including the second index
print("names[1:3] -- all names starting at position 1 up to and including position 3:\n", names[1:3],"\n")
````

## List of Lists

````python
list_of_lists = []
list1 = ['banana','yellow']
list2 = ['apple','red']
list_of_lists.append(list1)
list_of_lists.append(list2)

# Printed list of lists
print(list_of_lists)
print('-' * 50) # separation line

# * unpacks the list
print('Unpack lists into separate lists')
print(*list_of_lists, sep='\n')
print('-' * 50) # separation line

print('Name of 1st value in 1st list -> list_of_lists[0][0]: ',list_of_lists[0][0])
print('Name of 2nd value in 2nd list -> list_of_lists[1][1]:  ',list_of_lists[1][1])
````

## Dictionaries and Date Formatting
- Key Value Pairs in Curly Braces  
- Data indexed by Key  
- Current date formatting

````python
import datetime

# Date as a string
data1 = {'name':'bob','birthday':'2025-06-04'}
print('dictionary: ',data1,'\n')

# Creating date string from actual date time value
current_date = datetime.datetime.now()
print('current date value and type:',current_date,type(current_date),'\n')

# Format Year-Month-Day
current_date_as_string = current_date.strftime('%Y-%m-%d')
print('current date as string', current_date_as_string,'\n')

# Create a dictionary using just strings
data2 = {'name':'bob','birthday': current_date_as_string}
print('dictionary with date converted to string:',data2,'\n')

print(data1['name'], data2['birthday'])
````

## List of Dictionaries
Key/Value pairs indexed by key

````python
list_of_dictionaries = []
dict1 = {'Name':'Bob','Color':'blue'}
dict2 = {'Name': 'Sue', 'Color':'green'}
list_of_dictionaries.append(dict1)
list_of_dictionaries.append(dict2)

print(list_of_dictionaries)
print('-' * 50)


# * unpacks the list
print(*list_of_dictionaries, sep='\n')
print('-' * 50)

print('Name of 1st value in 1st dictionary: ',list_of_dictionaries[0]['Name'])
print('Name of 2nd value in 2nd dictionary: ',list_of_dictionaries[1]['Color'])
````

## CSV file format   
Comma Separated Value

Sample CSV Data: 

```csv
name,color  
bob,brown  
sue,blue  
jack,yellow  
mary,green  
sam,red  
ann,pink 
```

````python
with open('../../data/test.csv', 'r') as file:
    rows = file.readlines()
    print(rows, "\n")
    for row in rows:
        # strip off the new line for each row and split at comma into a list structure
        data = row.strip().split(',')
        print(data[0] + '\t' + data[1])
````
