# Pandas Example: Create, Load, Filter

```python
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Age": [25, 30, 35, 40],
    "City": ["Seattle", "Portland", "Seattle", "San Francisco"]
}
# load data into DataFrame
df = pd.DataFrame(data)
# print all data
print("\nAll Data\n")
print(df)
seattle_residents = df[df["City"] == "Seattle"]
# print filtered data
print("\nFiltered Data: Seattle Residents\n")
print(seattle_residents)
```
